```python
# Sample data
data = [
    (1, "John", "Kolkata", 25),
    (2, "Alice", "Delhi", 32),
    (3, "Bob", "Mumbai", 45),
    (4, "David", "Delhi", 28),
    (5, "Eva", "Kolkata", 39)
]

# Define schema
columns = ["customer_id", "name", "city", "age"]

# Create DataFrame
df = spark.createDataFrame(data, columns)

# Display in Databricks notebook
display(df)
```

```python
# Save DataFrame as a managed table in Databricks
df.write.saveAsTable("projects.default.customer_data")
```

```python
# New rows to add
new_data = [
    (6, "Frank", "Chennai", 41),
    (7, "Grace", "Pune", 29)
]

new_columns = ["customer_id", "name", "city", "age"]

# Create new DataFrame
df_new = spark.createDataFrame(new_data, new_columns)

# Append to existing table
df_new.write.mode("append").saveAsTable("projects.default.customer_data")
```

```python
# Sample order data
orders_data = [
    (101, 1, "Laptop", 75000),
    (102, 2, "Phone", 30000),
    (103, 3, "Tablet", 20000),
    (104, 5, "Headphones", 5000)
]

orders_columns = ["order_id", "customer_id", "product", "amount"]

# Create DataFrame
orders_df = spark.createDataFrame(orders_data, orders_columns)

# Save as a new table
orders_df.write.mode("overwrite").saveAsTable("projects.default.orders")
```
